```python
import os
import numpy as np

from pyFM.mesh import TriMesh
from tqdm.auto import tqdm

import meshplot as mp

def plot_mesh(myMesh,cmap=None):
    mp.plot(myMesh.vertlist, myMesh.facelist,c=cmap)
    
def double_plot(myMesh1,myMesh2,cmap1=None,cmap2=None):
    d = mp.subplot(myMesh1.vertlist, myMesh1.facelist, c=cmap1, s=[2, 2, 0])
    mp.subplot(myMesh2.vertlist, myMesh2.facelist, c=cmap2, s=[2, 2, 1], data=d)

def visu(vertices):
    min_coord,max_coord = np.min(vertices,axis=0,keepdims=True),np.max(vertices,axis=0,keepdims=True)
    cmap = (vertices-min_coord)/(max_coord-min_coord)
    return cmap
```

```python
meshlist = [TriMesh(f'data/shapes/shapes/{i:02d}.off', area_normalize=True, center=True).process(k=150, intrinsic=True) for i in range(381,401,1)]
```

```python
import os
import numpy as np

# Suppose your meshes are in meshlist[0], meshlist[1], ...
# and correspond to IDs 381, 382, ..., 400
mesh_ids = list(range(381, 401))  # Adjust to your actual IDs
meshes = {mesh_id: mesh for mesh_id, mesh in zip(mesh_ids, meshlist)}

map_folder = 'data/bim_maps/bim_maps'
map_files = os.listdir(map_folder)

for map_filename in map_files:
    if not map_filename.endswith(".map"):
        continue  # skip non-map files

    # Strip the .map extension
    name = map_filename[:-4]  # removes last 4 chars ".map"

    # Extract IDs from the filename
    id1_str, id2_str = name.split("_")
    id1, id2 = int(id1_str), int(id2_str)

    # Get meshes by ID
    if id1 not in meshes or id2 not in meshes:
        continue

    mesh1 = meshes[id1]
    mesh2 = meshes[id2]

    # Load the pointwise map and flatten
    p2p_21 = np.loadtxt(os.path.join(map_folder, map_filename), dtype=int).flatten()

    # Example condition (optional)
    if id1 < 385 and id2 < 385:
        cmap1 = visu(mesh1.vertlist); cmap2 = cmap1[p2p_21]
        double_plot(mesh1, mesh2, cmap1, cmap2)
```
